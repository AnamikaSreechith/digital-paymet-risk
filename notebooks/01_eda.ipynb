```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

# Project directories
ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

DATA_PATH = ROOT / "data" / "raw" / "creditcard.csv"
FIGURES_PATH = ROOT / "reports" / "figures"

FIGURES_PATH.mkdir(parents=True, exist_ok=True)

print("Project directory:", ROOT)
print("Dataset path:", DATA_PATH)
```

```python
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")

print("Shape:", df.shape)

df.head()
```

```python
# Dataset information
df.info()

# Statistical summary
display(df.describe())

# Column names
print(df.columns.tolist())
```

```python
print("Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDuplicate rows by class:")
print(df.loc[df.duplicated(), "Class"].value_counts())
```

```python
class_counts = df["Class"].value_counts()

print(class_counts)

print("\nPercentage Distribution:")
print(df["Class"].value_counts(normalize=True) * 100)

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Class")

plt.title("Genuine vs Fraudulent Transactions")
plt.xlabel("Transaction Class (0 = Genuine, 1 = Fraud)")
plt.ylabel("Number of Transactions")

plt.savefig(
    FIGURES_PATH / "class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Amount",
    bins=50,
    kde=True
)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.savefig(
    FIGURES_PATH / "amount_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Class",
    y="Amount"
)

plt.title("Transaction Amount by Class")
plt.xlabel("Class (0 = Genuine, 1 = Fraud)")
plt.ylabel("Transaction Amount")

plt.savefig(
    FIGURES_PATH / "amount_by_class.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
df["Time_Hours"] = df["Time"] / 3600

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Time_Hours",
    hue="Class",
    bins=50,
    element="step",
    stat="density",
    common_norm=False
)

plt.title("Transaction Distribution Over Time")
plt.xlabel("Elapsed Time (Hours)")
plt.ylabel("Density")

plt.savefig(
    FIGURES_PATH / "transaction_time.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Calculate correlation with target variable
class_corr = df.corr()["Class"].drop("Class")

# Sort correlations
class_corr = class_corr.sort_values()

plt.figure(figsize=(12, 8))

class_corr.plot(
    kind="barh",
    color=[
        "steelblue" if value < 0 else "coral"
        for value in class_corr
    ]
)

plt.title("Feature Correlation with Fraud Class")
plt.xlabel("Correlation Coefficient")
plt.ylabel("Features")

plt.tight_layout()

plt.savefig(
    FIGURES_PATH / "class_correlation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
selected_features = ["V1", "V2", "V3", "V4", "Amount"]

for feature in selected_features:

    plt.figure(figsize=(8, 4))

    sns.kdeplot(
        data=df,
        x=feature,
        hue="Class",
        common_norm=False
    )

    plt.title(f"{feature} Distribution by Transaction Class")

    plt.savefig(
        FIGURES_PATH / f"{feature}_distribution.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```
